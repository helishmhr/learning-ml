# **ML Pipeline**

## **Step 1: Data Collection**

- Gather data from sources like databases, APIs or CSV files.

## **Step 2: Data Exploration**

Exploratory Data Analysis (EDA) is a process of examining datasets to summarize their main characteristics, identify patterns, detect anomalies, and validate assumptions using visual and statistical techniques

## **Step 3: Data Preprocessing**

- Clean the data by handling missing values, duplicates and errors.
- Normalize and standardize numerical values.
- Convert categorical variables into a machine readable format.

## **Step 4: Feature Engineering**

- The process of selecting, transforming, or creating features that improve model performance.

## **Step 5: Data Splitting**

- Divide the dataset into training, validation and testing sets.
- When dealing with imbalanced datasets, use random sampling.

## **Step 6: Model Selection**

- Choose an algorithm appropriate for the task. Such as Classification, Regression, Clustering.
- Choosing the right model depends on various factors such as dataset size, complexity, accuracy requirements, computational cost and
interpretability.

## **Step 7: Hyperparameter Tuning**

- Hyperparameters control learning behavior.
- They are used to estimate the model parameters.
- They are set manually.
- The choice of hyperparameter decide how efficient the training is.

## **Step 8: Model Training**

- Training means teaching the model to use data.
- The objective is to minimize the prediction error

## **Step 9: Model Evaluation**

- Evaluate the performance of the model using various metrices such as Accuracy, Precision, F1 Score, Recall etc

## **Step 10: Model Deployment**

- After successful evaluation, the model is deployed for real-world use.
- Various deployment methods used are Web applications, Mobile applications, Cloud services, Embedded systems and APIs.

## **Step 11: Monitoring and Maintenance**

- Deployment is not the end.
- We need to monitor the model because data changes over time and this phenomenon is called data drift. Data drift is caused due to changes in input data distribution or changes in relationship between inputs and outputs.
- In case of change in data, we periodically retrain the model using new data.
