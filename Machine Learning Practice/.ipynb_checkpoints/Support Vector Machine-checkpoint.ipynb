# **Support Vector Machine With scikit-learn**

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
```

## **Linear SVM**

```python
# Training data: [hours studied, hours slept]
X = np.array([
    [1, 2],
    [2, 1],
    [2, 3],
    [3, 2],
    [6, 5],
    [7, 7],
    [8, 6],
    [9, 8]
])

# 0 = Fail, 1 = Pass
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])
```

```python
# Create the SVM model
model = SVC(kernel="linear",C=1.0)
```

```python
# Train the SVM model
model.fit(X, y)
```

```python
# Predict a new student's result
new_student = [[5, 4]]
prediction = model.predict(new_student)

print("Pass" if prediction[0] == 1 else "Fail")
```

```python
plt.scatter(
    X[:, 0], X[:, 1],
    c=y, cmap="bwr",
    edgecolors="k", s=100
)
# Plot the decision boundary
w = model.coef_[0]
b = model.intercept_[0]

x_values = np.linspace(0, 10, 100)
y_values = -(w[0] * x_values + b) / w[1]

plt.plot(x_values, y_values, "k-")
plt.xlabel("Hours Studied")
plt.ylabel("Hours Slept")
plt.title("Linear SVM Classification")
plt.show()
```

## **Non Linear SVM**

```python
from sklearn.datasets import make_circles
```

```python
# Training data

# Generate data
X, y = make_circles(
    n_samples=100,
    noise=0.05,
    factor=0.3,
    random_state=42
)
```

```python
# Create nonlinear SVM
model = SVC(kernel="rbf")
```

```python
# Train nonlinear SVM
model.fit(X, y)
```

```python
# Predict
print(model.predict([[0, 0]]))
```

```python
# Visualize data
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.xlabel("X1")
plt.ylabel("X2")
plt.title("Nonlinear SVM")
plt.show()
```
