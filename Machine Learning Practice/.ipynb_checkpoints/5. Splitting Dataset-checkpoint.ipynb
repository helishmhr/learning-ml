# **Splitting Dataset into Training, Validation & Test Sets**

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("insurance.csv")
df
```

```python
#separate input and target attributes (in this case "charges" is our target attribute)
X = df.drop("charges", axis=1)
Y = df["charges"]
```

```python
# First split: 80% training, 20% temporary data
X_train, X_temp, Y_train, Y_temp = train_test_split(
    X, Y, test_size=0.2, random_state=42
)
```

Here, random_state=42 ensures that you get the same train-test split every time you run the code, provided the input data and other settings remain unchanged.

```python
# Second split: divide the remaining 20% equally as Validation and Testing Set
X_val, X_test, Y_val, Y_test = train_test_split(X_temp, Y_temp,
    test_size=0.5,
    random_state=42,
)
```

```python
print("Training:", len(X_train))
print("Validation:", len(X_val))
print("Testing:", len(X_test))
```
