# **Linear Regression With scikit-learn**

## **Simple Linear Regression**

```python
import numpy as np
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
# Sample training data
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # tranforming 1D array into 2D because Scikit-learn's LinearRegression expects input features in a 2D format
y = np.array([2, 4, 5, 4, 5])
```

```python
# Create model instance
model = LinearRegression()
```

```python
# Fit the model
model.fit(x, y)
```

```python
# Get model parameters
print(f"Slope (B1): {model.coef_[0]}")
print(f"Intercept (B0): {model.intercept_}")
```

```python
# Predict values
y_pred = model.predict(x)
print("Predicted values:", y_pred)
```

```python
# Visualize actual data and regression line
plt.scatter(x, y, label="Actual data")
plt.plot(x, y_pred, color="red", label="Regression line")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Visualization")
plt.legend()
plt.show()
```

## **Multiple Linear Regression**

```python
#sample training data

# Input features: Area, Bedrooms, Age
X = np.array([
    [1000, 2, 10],
    [1500, 3, 5],
    [1800, 3, 2],
    [2000, 4, 5],
    [2500, 4, 1],
    [3000, 5, 3]
])

# Target: House price in lakhs
Y = np.array([50, 75, 95, 100, 130, 150])
```

```python
# Create the model
model = LinearRegression()
```

```python
# Train the model
model.fit(X, y)
```

```python
# Display the learned parameters
print("Intercept (b0):", model.intercept_)
print("Coefficients (b1, b2, b3):", model.coef_)
```

```python
# Predict the price of a new house
# Area = 2200 sq. ft., Bedrooms = 3, Age = 4 years
new_house = np.array([[2200, 3, 4]])
predicted_price = model.predict(new_house)
print("Predicted house price:", predicted_price[0], "lakhs")
```

## **Polynomial Regression**
Implementing polynomial regression with scikit-learn is very similar to linear regression. There’s only one extra step: you need to transform the array of inputs to include nonlinear terms such as 𝑥².

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
# Sample training data
X = np.array([5, 15, 25, 35, 45, 55]).reshape((-1, 1))
Y = np.array([15, 11, 2, 8, 25, 32])
```

```python
#transform the input array x to contain any additional columns with the values of 𝑥², and eventually more features.
transformer = PolynomialFeatures(degree=2,include_bias=False)
transformer.fit(X)
X_poly = transformer.transform(X)
```

This would have transformed our input into three columns:
$[1, x, x^2]$ but setting *include_bias=False*  transforms your input into 2 columns $[x, x^2]$

```python
print("Original X:")
print(X)

print("Transformed X:")
print(X_poly)
```

```python
#create the model
model = LinearRegression()
```

```python
#train the model
model.fit(X_poly, Y)
```

```python
# Display the learned parameters
print("Intercept (B0):", model.intercept_)
print("Coefficients (B1, B2):", model.coef_)
```

```python
#Predict
Y_pred = model.predict(X_poly)
print("Predictions:", Y_pred)
```

```python
#Visualizing
plt.scatter(X, Y, label="Actual data")
plt.plot(X, Y_pred, color="red", label="Polynomial curve")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

## **Regression Evaluation Metrics**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score
```

```python
y_actual = [3, -0.5, 2, 7]
y_pred   = [2.5, 0.0, 2, 8]
```

```python
#MAE
print(mean_absolute_error(y_actual, y_pred))
```

```python
#MSE
print(mean_squared_error(y_actual, y_pred))
```

```python
#RSME
print(root_mean_squared_error(y_actual, y_pred))
```

```python
#R-Squared
print(r2_score(y_actual, y_pred))
```
